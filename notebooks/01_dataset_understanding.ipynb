```python
import numpy as np
import pandas as pd

# Input features: exclude both target columns
X = df_model.drop(
    columns=["Attack_type", "Binary_target"]
).copy()

# Two prediction targets
y_binary = df_model["Binary_target"].copy()
y_multi = df_model["Attack_type"].copy()

# Remove the constant feature identified during EDA
constant_features = ["bwd_URG_flag_count"]

X = X.drop(columns=constant_features)

print("Input feature shape:", X.shape)
print("Binary target shape:", y_binary.shape)
print("Multiclass target shape:", y_multi.shape)

print("\nRemaining constant columns:")
print(X.nunique()[X.nunique() <= 1])
```

```python
# Generate a consistent hash for each feature combination
groups = pd.util.hash_pandas_object(
    X,
    index=False
).to_numpy()

print("Total observations:", len(X))
print("Unique feature groups:", len(np.unique(groups)))
print("Repeated feature observations:", 
      len(X) - len(np.unique(groups)))
```

```python
from sklearn.model_selection import StratifiedGroupKFold

# First split: 80% development pool, 20% test
sgkf_test = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

train_val_idx, test_idx = next(
    sgkf_test.split(
        X,
        y_multi,
        groups=groups
    )
)

print("Development pool:", len(train_val_idx))
print("Test set:", len(test_idx))
```

```python
# Verify all multiclass labels are represented in every split

all_classes = set(y_multi.unique())

for name, labels in {
    "Train": y_multi_train,
    "Validation": y_multi_val,
    "Test": y_multi_test
}.items():

    missing_classes = all_classes - set(labels.unique())

    print(f"{name}:")
    print("Total classes:", labels.nunique())
    print("Missing classes:", missing_classes)

    assert not missing_classes, (
        f"{name} is missing classes: {missing_classes}"
    )

print("All classes are represented in every split.")
```

```python
# Extract the development pool
X_train_val = X.iloc[train_val_idx]
y_train_val = y_multi.iloc[train_val_idx]
groups_train_val = groups[train_val_idx]

# Split development pool into approximately 80% train and 20% validation
sgkf_val = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

train_relative_idx, val_relative_idx = next(
    sgkf_val.split(
        X_train_val,
        y_train_val,
        groups=groups_train_val
    )
)

# Convert relative indices into original dataframe indices
train_idx = train_val_idx[train_relative_idx]
val_idx = train_val_idx[val_relative_idx]

# Final datasets
X_train = X.iloc[train_idx].copy()
X_val = X.iloc[val_idx].copy()
X_test = X.iloc[test_idx].copy()

y_binary_train = y_binary.iloc[train_idx].copy()
y_binary_val = y_binary.iloc[val_idx].copy()
y_binary_test = y_binary.iloc[test_idx].copy()

y_multi_train = y_multi.iloc[train_idx].copy()
y_multi_val = y_multi.iloc[val_idx].copy()
y_multi_test = y_multi.iloc[test_idx].copy()

print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Testing:", X_test.shape)
```

```python
train_groups = set(groups[train_idx])
val_groups = set(groups[val_idx])
test_groups = set(groups[test_idx])

print("Train-Val overlap:", len(train_groups & val_groups))
print("Train-Test overlap:", len(train_groups & test_groups))
print("Val-Test overlap:", len(val_groups & test_groups))

assert len(train_groups & val_groups) == 0
assert len(train_groups & test_groups) == 0
assert len(val_groups & test_groups) == 0

print("\nNo identical feature groups cross the splits!")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)
from sklearn.impute import SimpleImputer

# Numerical preprocessing
numerical_pipeline_scaled = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

numerical_pipeline_unscaled = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore"
    ))
])

# Preprocessing for Logistic Regression, KNN and SVM
preprocessor_scaled = ColumnTransformer([
    (
        "numerical",
        numerical_pipeline_scaled,
        numerical_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])

# Preprocessing for tree-based models
preprocessor_tree = ColumnTransformer([
    (
        "numerical",
        numerical_pipeline_unscaled,
        numerical_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])
```
